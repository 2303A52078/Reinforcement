```python
# Greedy Multi-Armed Bandit using Arithmetic Mean

n = int(input("Enter number of machines: "))

machines = {}

for i in range(n):
    name = input(f"\nEnter machine name {i + 1}: ")
    rewards = list(map(float, input("Enter rewards separated by spaces: ").split()))
    machines[name] = rewards

q_values = {}

for name, rewards in machines.items():
    q_values[name] = sum(rewards) / len(rewards)

    print(f"\nMachine {name}")
    print(f"Rewards: {rewards}")
    print(f"Q({name}) = {q_values[name]:.2f}")

selected_machine = max(q_values, key=q_values.get)

print(f"\nGreedy Selected Machine: {selected_machine}")
```

```python
# Greedy Multi-Armed Bandit using Incremental Update

n = int(input("Enter number of machines: "))

machines = {}

# Take machine names and rewards
for i in range(n):
    name = input(f"\nEnter machine name {i + 1}: ")
    rewards = list(map(float, input("Enter rewards separated by spaces: ").split()))
    machines[name] = rewards

# Initialize Q and N
Q = {}
N = {}

for name in machines:
    Q[name] = 0
    N[name] = 0

# Process each reward
for name, rewards in machines.items():

    print(f"\nMachine {name}")

    for reward in rewards:
        # Increment number of plays
        N[name] += 1

        # Incremental update
        Q[name] = Q[name] + (reward - Q[name]) / N[name]

        print(
            f"Reward = {reward}, "
            f"N({name}) = {N[name]}, "
            f"Q({name}) = {Q[name]:.2f}"
        )

# Display final Q values
print("\nFinal Estimated Q Values:")

for name in machines:
    print(f"Q({name}) = {Q[name]:.2f}")

# Greedy selection
selected_machine = max(Q, key=Q.get)

print(f"\nGreedy Selected Machine: {selected_machine}")
```
